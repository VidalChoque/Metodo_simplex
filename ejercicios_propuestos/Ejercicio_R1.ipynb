```python
# =====================================================================
# EJERCICIO R1 (REPLICAR): Maximización con 3 variables
# =====================================================================
# PROBLEMA:
#   Tres artículos: X (x1), Y (x2), Z (x3).
#   Ganancias: $20, $30, $40.
#   Restricciones:
#     2x1 + 3x2 + 4x3 <= 60  (horas máquina)
#     x1  + 2x2 + x3  <= 40  (materia prima)
#
# OBJETIVO: maximizar ganancia.
# =====================================================================

import numpy as np
from scipy.optimize import linprog

print("="*70)
print("EJERCICIO R1 (REPLICAR): MAXIMIZACIÓN CON 3 VARIABLES")
print("="*70)

# ---------------------------------------------------------------------
# PASO 1: MODELO
# ---------------------------------------------------------------------
print("\n--- PASO 1: MODELO ---")
print("max z = 20*x1 + 30*x2 + 40*x3")
print("s.a.")
print("  2*x1 + 3*x2 + 4*x3 <= 60")
print("  1*x1 + 2*x2 + 1*x3 <= 40")
print("  x1, x2, x3 >= 0")

# ---------------------------------------------------------------------
# PASO 2: FORMA ESTÁNDAR
# ---------------------------------------------------------------------
print("\n--- PASO 2: FORMA ESTÁNDAR ---")
print("  2*x1 + 3*x2 + 4*x3 + s1 = 60")
print("  1*x1 + 2*x2 + 1*x3 + s2 = 40")
print("  z - 20*x1 - 30*x2 - 40*x3 = 0")

# ---------------------------------------------------------------------
# PASO 3: SOLUCIÓN CON LINPROG
# ---------------------------------------------------------------------
print("\n--- PASO 3: SOLUCIÓN CON SCIPY ---")
c = [-20, -30, -40]
A_ub = [[2, 3, 4], [1, 2, 1]]
b_ub = [60, 40]
bounds = [(0, None)] * 3

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

print(f"  x1 = {res.x[0]:.4f}")
print(f"  x2 = {res.x[1]:.4f}")
print(f"  x3 = {res.x[2]:.4f}")
print(f"  z  = {-res.fun:.4f}")

# ---------------------------------------------------------------------
# PASO 4: VERIFICACIÓN MANUAL
# ---------------------------------------------------------------------
print("\n--- PASO 4: VERIFICACIÓN MANUAL ---")
print(f"  Restricción 1: 2({res.x[0]:.1f}) + 3({res.x[1]:.1f}) + "
      f"4({res.x[2]:.1f}) = {2*res.x[0]+3*res.x[1]+4*res.x[2]:.2f} <= 60")
print(f"  Restricción 2: 1({res.x[0]:.1f}) + 2({res.x[1]:.1f}) + "
      f"1({res.x[2]:.1f}) = {res.x[0]+2*res.x[1]+res.x[2]:.2f} <= 40")
print(f"  Ganancia: 20({res.x[0]:.1f}) + 30({res.x[1]:.1f}) + "
      f"40({res.x[2]:.1f}) = {-res.fun:.2f}")

# ---------------------------------------------------------------------
# PASO 5: RESPUESTA ESPERADA
# ---------------------------------------------------------------------
print("\n--- PASO 5: RESPUESTA ESPERADA ---")
print("  x1 = 0, x2 = 0, x3 = 15, z = 600")
print("  Verificación: 4(15)=60 <= 60, 1(15)=15 <= 40")

# ---------------------------------------------------------------------
# PASO 6: ANÁLISIS
# ---------------------------------------------------------------------
print("\n--- PASO 6: ANÁLISIS ---")
print("1. Conviene producir solo el artículo Z (el de mayor ganancia).")
print("2. La restricción de horas máquina se agota (cuello de botella).")
print("3. La materia prima tiene holgura de 25 unidades.")
print("4. El precio sombra de las horas máquina es 10.")
print("5. Este ejercicio muestra que el óptimo puede estar en un")
print("   vértice donde solo una variable es positiva.")
print("="*70)
```
