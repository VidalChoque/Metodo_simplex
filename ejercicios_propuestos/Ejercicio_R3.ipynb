```python
# =====================================================================
# EJERCICIO R3 (REPLICAR): Minimización con restricciones >=
# =====================================================================
# PROBLEMA:
#   Refinería: gasolina (x1) y diésel (x2).
#   Costos: $5/unidad gasolina, $8/unidad diésel.
#   Restricciones:
#     x1 + x2 >= 30  (demanda total mínima)
#     x1       >= 20  (gasolina mínima)
#     x2       >= 15  (diésel mínimo)
#
# OBJETIVO: minimizar costo.
# =====================================================================

import numpy as np
from scipy.optimize import linprog

print("="*70)
print("EJERCICIO R3 (REPLICAR): MINIMIZACIÓN CON RESTRICCIONES >=")
print("="*70)

# ---------------------------------------------------------------------
# PASO 1: MODELO
# ---------------------------------------------------------------------
print("\n--- PASO 1: MODELO ---")
print("min z = 5*x1 + 8*x2")
print("s.a.")
print("  x1 + x2 >= 30")
print("  x1      >= 20")
print("  x2      >= 15")
print("  x1, x2 >= 0")

# ---------------------------------------------------------------------
# PASO 2: FORMA ESTÁNDAR (convertir >= a <=)
# ---------------------------------------------------------------------
print("\n--- PASO 2: FORMA ESTÁNDAR ---")
print("Se multiplican las restricciones >= por -1:")
print("  -x1 - x2 <= -30")
print("  -x1      <= -20")
print("  -x2      <= -15")
print("Se agregan excesos e1, e2, e3 (y artificiales para Fase I)")

# ---------------------------------------------------------------------
# PASO 3: SOLUCIÓN CON LINPROG
# ---------------------------------------------------------------------
print("\n--- PASO 3: SOLUCIÓN CON SCIPY ---")
c = [5, 8]
A_ub = [[-1, -1], [-1, 0], [0, -1]]
b_ub = [-30, -20, -15]
bounds = [(0, None), (0, None)]

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

print(f"  x1 = {res.x[0]:.4f}")
print(f"  x2 = {res.x[1]:.4f}")
print(f"  costo = {res.fun:.4f}")

# ---------------------------------------------------------------------
# PASO 4: VERIFICACIÓN MANUAL
# ---------------------------------------------------------------------
print("\n--- PASO 4: VERIFICACIÓN MANUAL ---")
print(f"  x1 + x2 = {res.x[0]:.1f} + {res.x[1]:.1f} = "
      f"{res.x[0]+res.x[1]:.1f} >= 30")
print(f"  x1 = {res.x[0]:.1f} >= 20")
print(f"  x2 = {res.x[1]:.1f} >= 15")
print(f"  Costo: 5({res.x[0]:.1f}) + 8({res.x[1]:.1f}) = {res.fun:.2f}")

# ---------------------------------------------------------------------
# PASO 5: RESPUESTA ESPERADA
# ---------------------------------------------------------------------
print("\n--- PASO 5: RESPUESTA ESPERADA ---")
print("  x1 = 20, x2 = 15, costo = 220")
print("  Verificación: 20+15=35 >= 30, 20 >= 20, 15 >= 15")
print("  Costo: 5(20) + 8(15) = 100 + 120 = 220")

# ---------------------------------------------------------------------
# PASO 6: ANÁLISIS
# ---------------------------------------------------------------------
print("\n--- PASO 6: ANÁLISIS ---")
print("1. Producir 20 unidades de gasolina y 15 de diésel.")
print("2. Costo mínimo: $220.")
print("3. La demanda total (30) se cumple con holgura (35).")
print("4. Las restricciones individuales están activas.")
print("5. Este ejercicio usa Fase I porque hay restricciones >=.")
print("6. El precio sombra de la demanda total es 0 (ya se cumple).")
print("="*70)
```
